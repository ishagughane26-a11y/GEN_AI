```python
import streamlit as st
import nltk
import spacy
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer


nltk.download('wordnet')
nltk.download('omw-1.4')

# ------------------------------------------------------------------
# Page config
# ------------------------------------------------------------------
st.set_page_config(
    page_title="TextSense AI - NLP Workbench",
    page_icon="🧠",
    layout="wide",
    initial_sidebar_state="expanded",
)

st.markdown("""
<style>
@import url('https://fonts.googleapis.com/css2?family=DM+Sans:wght@400;500;600;700;800&display=swap');

html, body, [class*="css"] {
    font-family: 'DM Sans', sans-serif;
}

:root {
    --ink: #172033;
    --muted: #687386;
    --paper: #fbfaf7;
    --panel: #ffffff;
    --line: #e7e2d9;
    --violet: #6d4aff;
    --violet-soft: #f0edff;
    --mint: #dff7ed;
    --mint-ink: #167052;
}

.stApp {
    background:
        radial-gradient(circle at 90% 0%, #eeeaff 0, transparent 28%),
        #fbfaf7;
    color: var(--ink);
}

.block-container {
    max-width: 1220px;
    padding: 2.2rem 2.8rem 3rem;
}

p, span, label, li, div {
    color: var(--ink);
}

/* Header */
.ts-hero {
    position: relative;
    overflow: hidden;
    padding: 2.2rem 2.4rem;
    border-radius: 26px;
    background: #ffffff;
    border: 1px solid var(--line);
    box-shadow: 0 16px 40px rgba(35, 28, 70, .08);
    margin-bottom: 1.5rem;
}

.ts-hero:before {
    content: "";
    position: absolute;
    width: 170px;
    height: 170px;
    right: -55px;
    top: -65px;
    border-radius: 50%;
    background: #e9e4ff;
}

.ts-hero h1 {
    position: relative;
    margin: 0;
    color: #17122f !important;
    font-size: 2.6rem;
    font-weight: 800;
    letter-spacing: -.045em;
}

.ts-hero p {
    position: relative;
    margin: .55rem 0 0;
    max-width: 680px;
    color: var(--muted);
    font-size: 1rem;
}

.ts-pills {
    position: relative;
    display: flex;
    gap: .5rem;
    margin-top: 1rem;
    flex-wrap: wrap;
}

.ts-pill {
    padding: .34rem .75rem;
    border-radius: 999px;
    background: var(--violet-soft);
    border: 1px solid #ded7ff;
    color: #5137c9;
    font-size: .78rem;
    font-weight: 700;
}

/* Sidebar */
section[data-testid="stSidebar"] {
    background: #17152a;
    border-right: none;
}

section[data-testid="stSidebar"] * {
    color: #f7f5ff !important;
}

section[data-testid="stSidebar"] h1 {
    font-size: 1.45rem;
    font-weight: 800;
    padding-bottom: .7rem;
    border-bottom: 1px solid #34304e;
}

section[data-testid="stSidebar"] textarea {
    background: #211e39;
    border: 1px solid #3c375d;
    border-radius: 14px;
    color: #fff;
}

section[data-testid="stSidebar"] textarea:focus {
    border-color: #8d78ff;
    box-shadow: 0 0 0 1px #8d78ff;
}

section[data-testid="stSidebar"] div[data-baseweb="select"] > div {
    background: #211e39;
    border-color: #3c375d;
    border-radius: 12px;
}

section[data-testid="stSidebar"] hr {
    border-color: #34304e;
}

/* Primary action */
.stButton > button {
    min-height: 48px;
    border: 0;
    border-radius: 13px;
    background: #6d4aff;
    color: white;
    font-weight: 800;
    box-shadow: 0 10px 22px rgba(109, 74, 255, .25);
}

.stButton > button:hover {
    background: #5837e8;
    color: white;
    transform: translateY(-1px);
}

/* Stat cards */
.stat {
    background: white;
    border: 1px solid var(--line);
    border-radius: 18px;
    padding: 1.1rem 1.25rem;
    box-shadow: 0 7px 20px rgba(35, 28, 70, .055);
}

.stat-label {
    color: #8a8490;
    font-size: .72rem;
    font-weight: 800;
    text-transform: uppercase;
    letter-spacing: .08em;
}

.stat-number {
    color: #201a3b;
    font-size: 1.8rem;
    font-weight: 800;
    margin-top: .15rem;
}

/* Content cards */
div[data-testid="stExpander"],
div[data-testid="stVerticalBlockBorderWrapper"] {
    border: 1px solid var(--line);
    border-radius: 18px;
    background: white;
    box-shadow: 0 7px 20px rgba(35, 28, 70, .055);
}

.section-heading {
    display: flex;
    align-items: center;
    gap: .6rem;
    color: #201a3b;
    font-size: 1.25rem;
    font-weight: 800;
    margin: .2rem 0 1rem;
}

.section-heading span {
    display: inline-flex;
    width: 30px;
    height: 30px;
    align-items: center;
    justify-content: center;
    border-radius: 9px;
    background: var(--violet-soft);
    color: var(--violet);
    font-size: .82rem;
}

.chip-row {
    display: flex;
    flex-wrap: wrap;
    gap: .45rem;
    padding: .8rem 0;
}

.chip {
    display: inline-block;
    padding: .36rem .68rem;
    border-radius: 9px;
    background: #f4f2ff;
    border: 1px solid #ddd7ff;
    color: #4d3bb1;
    font-size: .86rem;
    font-weight: 600;
}

.chip-muted {
    background: #f4f7f6;
    border-color: #dbe8e2;
    color: #397260;
}

.sentence-card {
    padding: .85rem 1rem;
    margin: .55rem 0;
    border: 1px solid var(--line);
    border-left: 4px solid var(--violet);
    border-radius: 12px;
    background: #fcfbff;
}

.sentence-card b {
    color: var(--violet);
    margin-right: .5rem;
}

.noun-chunk-chip {
    display: inline-block;
    margin: .25rem .4rem .25rem 0;
    padding: .4rem .75rem;
    border-radius: 999px;
    background: var(--mint);
    border: 1px solid #bcebd9;
    color: var(--mint-ink);
    font-size: .86rem;
    font-weight: 700;
}

[data-testid="stDataFrame"] {
    border: 1px solid var(--line);
    border-radius: 14px;
    overflow: hidden;
}

[data-testid="stAlert"] {
    border-radius: 13px;
}

h1, h2, h3 {
    color: var(--ink);
}
</style>
""", unsafe_allow_html=True)


# ------------------------------------------------------------------
# One-time downloads / model loading (cached so it only runs once)
# ------------------------------------------------------------------
@st.cache_resource
def load_resources():
    for pkg in ["punkt_tab", "stopwords", "wordnet", "averaged_perceptron_tagger_eng"]:
        try:
            nltk.download(pkg, quiet=True)
        except Exception:
            pass
    nlp_model = spacy.load("en_core_web_sm")
    return nlp_model


with st.spinner("Loading NLP models (first run only)..."):
    nlp = load_resources()

stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# ------------------------------------------------------------------
# Sidebar — text input + step selector
# ------------------------------------------------------------------
st.sidebar.title("🧠 TextSense Workspace")

default_text = (
    "Apple Inc. is planning to open a new office in Bangalore next year. "
    "Tim Cook announced this during a press conference in California."
)

st.sidebar.markdown("Input Text")
text = st.sidebar.text_area("📝 Enter Text", value=default_text, height=180, label_visibility="collapsed")

steps = [
    "1. Sentence Segmentation",
    "2. Word Tokenization",
    "3. Stop Word Removal",
    "4. Stemming",
    "5. Lemmatization",
    "6. POS Tagging",
    "7. Named Entity Recognition (NER)",
    "8. Dependency Parsing",
    "9. Noun Phrase Chunking",
    "🔁 Run All Steps",
]

st.sidebar.markdown("Pipeline Step")
selected_step = st.sidebar.selectbox(
    "Analysis module",
    steps,
    index=len(steps) - 1,
    help="Choose one transformation or run the complete pipeline."
)

st.sidebar.markdown("---")
run_button = st.sidebar.button("⚡ Analyze Text", type="primary", use_container_width=True)

# ------------------------------------------------------------------
# Main area
# ------------------------------------------------------------------
st.markdown("""
<div class="nlp-hero">
    <h1>✦ TextSense AI</h1>
    <h2>Interactive NLP Workbench
    <p>Explore a full interactive language intelligence workspace — from tokenization to dependency
    parsing — powered by <b>NLTK</b> &amp; <b>spaCy</b>. Pick a step in the control panel, drop in your text,
    and run the analysis.</p>
    <div class="badge-row">
        <span class="nlp-badge">NLTK</span>
        <span class="nlp-badge">spaCy</span>
        <span class="nlp-badge">9 Pipeline Steps</span>
    </div>
</div>
""", unsafe_allow_html=True)

with st.expander("📝 Working Text", expanded=True):
    if text.strip():
        st.markdown(
            f'<div style="padding:1rem;border:1px solid #e7e2d9;border-radius:14px;background:#fff;">{text}</div>',
            unsafe_allow_html=True
        )
    else:
        st.info("Add some text from the workspace panel to begin.")

st.markdown("### ✦ Analysis workspace")


# ------------------------------------------------------------------
# Step functions
# ------------------------------------------------------------------
def step_sentence_segmentation(text):
    st.markdown('<div class="section-heading"><span>01</span>Sentence Segmentation</div>', unsafe_allow_html=True)
    sentences = sent_tokenize(text)
    for i, sent in enumerate(sentences, 1):
        st.markdown(f'<div class="sentence-card"><b>{i}</b>{sent}</div>', unsafe_allow_html=True)
    return sentences


def step_word_tokenization(text):
    st.markdown('<div class="section-heading"><span>02</span>Word Tokenization</div>', unsafe_allow_html=True)
    words = word_tokenize(text)
    chips = "".join(f'<span class="chip">{w}</span>' for w in words)
    st.markdown(f'<div class="chip-row">{chips}</div>', unsafe_allow_html=True)
    return words


def step_stopword_removal(words):
    st.markdown('<div class="section-heading"><span>03</span>Stop Word Removal</div>', unsafe_allow_html=True)
    filtered_words = [w for w in words if w.lower() not in stop_words]
    chips = "".join(f'<span class="chip">{w}</span>' for w in filtered_words)
    st.markdown(f'<div class="chip-row">{chips}</div>', unsafe_allow_html=True)
    return filtered_words


def step_stemming(filtered_words):
    st.markdown('<div class="section-heading"><span>04</span>Stemming</div>', unsafe_allow_html=True)
    stemmed_words = [stemmer.stem(w) for w in filtered_words]
    chips = "".join(f'<span class="chip chip-muted">{w}</span>' for w in stemmed_words)
    st.markdown(f'<div class="chip-row">{chips}</div>', unsafe_allow_html=True)
    return stemmed_words


def step_lemmatization(filtered_words):
    st.markdown('<div class="section-heading"><span>05</span>Lemmatization</div>', unsafe_allow_html=True)
    lemmatized_words = [lemmatizer.lemmatize(w) for w in filtered_words]
    chips = "".join(f'<span class="chip chip-muted">{w}</span>' for w in lemmatized_words)
    st.markdown(f'<div class="chip-row">{chips}</div>', unsafe_allow_html=True)
    return lemmatized_words


def step_pos_tagging(words):
    st.markdown('<div class="section-heading"><span>06</span>POS Tagging</div>', unsafe_allow_html=True)
    pos_tags = nltk.pos_tag(words)
    st.dataframe(
        {"Word": [w for w, t in pos_tags], "POS Tag": [t for w, t in pos_tags]},
        use_container_width=True,
    )
    return pos_tags


def step_ner(doc):
    st.markdown('<div class="section-heading"><span>07</span>Named Entity Recognition</div>', unsafe_allow_html=True)
    if doc.ents:
        st.dataframe(
            {"Entity": [ent.text for ent in doc.ents], "Label": [ent.label_ for ent in doc.ents]},
            use_container_width=True,
        )
    else:
        st.info("No named entities found.")


def step_dependency_parsing(doc):
    st.markdown('<div class="section-heading"><span>08</span>Dependency Parsing</div>', unsafe_allow_html=True)
    st.dataframe(
        {
            "Token": [tok.text for tok in doc],
            "Dependency": [tok.dep_ for tok in doc],
            "Head": [tok.head.text for tok in doc],
        },
        use_container_width=True,
    )
    try:
        from spacy import displacy

        html = displacy.render(doc, style="dep", options={"compact": True})
        st.components.v1.html(html, height=400, scrolling=True)
    except Exception:
        pass


def step_noun_chunks(doc):
    st.markdown('<div class="section-heading"><span>09</span>Noun Phrase Chunking</div>', unsafe_allow_html=True)
    chunks = list(doc.noun_chunks)
    if chunks:
        chips = "".join(f'<span class="noun-chunk-chip">{chunk.text}</span>' for chunk in chunks)
        st.markdown(chips, unsafe_allow_html=True)
    else:
        st.info("No noun phrase chunks found.")


# ------------------------------------------------------------------
# Execution
# ------------------------------------------------------------------
if run_button:
    if not text.strip():
        st.warning("Please enter some text in the sidebar first.")
    else:
        words = word_tokenize(text)
        doc = nlp(text)

        with st.container(border=True):
            if selected_step == steps[0]:
                step_sentence_segmentation(text)

            elif selected_step == steps[1]:
                step_word_tokenization(text)

            elif selected_step == steps[2]:
                step_stopword_removal(words)

            elif selected_step == steps[3]:
                filtered = [w for w in words if w.lower() not in stop_words]
                step_stemming(filtered)

            elif selected_step == steps[4]:
                filtered = [w for w in words if w.lower() not in stop_words]
                step_lemmatization(filtered)

            elif selected_step == steps[5]:
                step_pos_tagging(words)

            elif selected_step == steps[6]:
                step_ner(doc)

            elif selected_step == steps[7]:
                step_dependency_parsing(doc)

            elif selected_step == steps[8]:
                step_noun_chunks(doc)

            elif selected_step == steps[9]:  # Run All
                step_sentence_segmentation(text)
                st.divider()
                step_word_tokenization(text)
                st.divider()
                filtered = step_stopword_removal(words)
                st.divider()
                step_stemming(filtered)
                st.divider()
                step_lemmatization(filtered)
                st.divider()
                step_pos_tagging(words)
                st.divider()
                step_ner(doc)
                st.divider()
                step_dependency_parsing(doc)
                st.divider()
                step_noun_chunks(doc)
else:
    st.info("👈 Enter text and click **Run Analysis** in the control panel to see results.")

#===============commands to run==============
#pip install streamlit nltk spacy
#python -m spacy download en_core_web_sm
#streamlit run nlp_app.py
```
